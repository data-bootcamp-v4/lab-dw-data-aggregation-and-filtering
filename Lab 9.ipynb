```python
# 1. Make a copy
clean_customer_marketing_df = customer_marketing_df.copy()
```

```python
clean_customer_marketing_df.duplicated().sum()
```

```python
clean_customer_marketing_df = (
    clean_customer_marketing_df
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
clean_customer_marketing_df.duplicated().sum()
```

```python
clean_customer_marketing_df.columns = (
    clean_customer_marketing_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)
```

```python
clean_customer_marketing_df.isnull().sum()
```

```python
clean_customer_marketing_df = clean_customer_marketing_df.drop(
    columns=["vehicle_type"]
)
```

```python
numeric_columns_with_missing_values = [
    "months_since_last_claim",
    "number_of_open_complaints"
]

for column in numeric_columns_with_missing_values:
    clean_customer_marketing_df[column] = (
        clean_customer_marketing_df[column]
        .fillna(clean_customer_marketing_df[column].median())
    )
```

```python
categorical_columns_with_missing_values = [
    "state",
    "response"
]

for column in categorical_columns_with_missing_values:
    clean_customer_marketing_df[column] = (
        clean_customer_marketing_df[column]
        .fillna(clean_customer_marketing_df[column].mode()[0])
    )
```

```python
clean_customer_marketing_df.info()
```

```python
clean_customer_marketing_df = clean_customer_marketing_df.drop(
    columns=["unnamed:_0"]
)
```

```python
clean_customer_marketing_df.isnull().sum()
```

```python
low_claim_df = clean_customer_marketing_df[
    (clean_customer_marketing_df["total_claim_amount"] < 1000) &
    (clean_customer_marketing_df["response"] == "Yes")
]

low_claim_df.head()
```

```python
low_claim_df.shape
```

```python
yes_response_df = clean_customer_marketing_df[
    clean_customer_marketing_df["response"] == "Yes"
].copy()

yes_response_df.shape
```

```python
response_analysis_df = (
    yes_response_df
    .groupby(["policy_type", "gender"], as_index=False)
    .agg(
        average_monthly_premium=(
            "monthly_premium_auto", "mean"
        ),
        average_customer_lifetime_value=(
            "customer_lifetime_value", "mean"
        ),
        average_total_claim_amount=(
            "total_claim_amount", "mean"
        ),
        number_of_customers=(
            "customer", "count"
        )
    )
    .round(2)
)

response_analysis_df
```

```python
response_analysis_df.sort_values(
    by="average_customer_lifetime_value",
    ascending=False
)
```

```python
customers_by_state_df = (
    customer_marketing_df
    .dropna(subset=["State"])
    .groupby("State", as_index=False)
    .agg(
        total_customers=("Customer", "nunique")
    )
    .sort_values(
        by="total_customers",
        ascending=False
    )
)

customers_by_state_df
```

```python
states_above_500_df = customers_by_state_df[
    customers_by_state_df["total_customers"] > 500
].reset_index(drop=True)

states_above_500_df
```

```python
clv_by_education_gender_df = (
    clean_customer_marketing_df
    .groupby(["education", "gender"], as_index=False)
    .agg(
        maximum_customer_lifetime_value=(
            "customer_lifetime_value", "max"
        ),
        minimum_customer_lifetime_value=(
            "customer_lifetime_value", "min"
        ),
        median_customer_lifetime_value=(
            "customer_lifetime_value", "median"
        )
    )
    .round(2)
)

clv_by_education_gender_df
```

```python
clv_by_education_gender_df.sort_values(
    by="median_customer_lifetime_value",
    ascending=False
)
```

```python
clean_customer_marketing_df["effective_to_date"] = pd.to_datetime(
    clean_customer_marketing_df["effective_to_date"],
    format="%m/%d/%y"
)

clean_customer_marketing_df["month"] = (
    clean_customer_marketing_df["effective_to_date"].dt.month
)
```

```python
clean_customer_marketing_df[
    ["effective_to_date", "month"]
].head()
```

```python
policies_by_state_month_df = (
    clean_customer_marketing_df
    .groupby(["state", "month"])
    .size()
    .reset_index(name="policies_sold")
)

policies_by_state_month_df.head()
```

```python
policies_state_month_pivot_df = (
    policies_by_state_month_df
    .pivot(
        index="state",
        columns="month",
        values="policies_sold"
    )
    .fillna(0)
    .astype(int)
)

policies_state_month_pivot_df
```

```python
total_policies_by_state_df = (
    policies_by_state_month_df
    .groupby("state", as_index=False)["policies_sold"]
    .sum()
    .sort_values(
        by="policies_sold",
        ascending=False
    )
)

total_policies_by_state_df
```

```python
top_3_states = (
    total_policies_by_state_df
    .head(3)["state"]
    .tolist()
)

top_3_states
```

```python
top_3_states_monthly_df = policies_by_state_month_df[
    policies_by_state_month_df["state"].isin(top_3_states)
]

top_3_states_monthly_df
```

```python
top_3_states_pivot_df = (
    top_3_states_monthly_df
    .pivot(
        index="state",
        columns="month",
        values="policies_sold"
    )
    .fillna(0)
    .astype(int)
)

top_3_states_pivot_df
```

```python
top_3_states_pivot_df = top_3_states_pivot_df.rename(
    columns={
        1: "January",
        2: "February"
    }
)

top_3_states_pivot_df
```

```python
channel_response_wide_df = pd.crosstab(
    clean_customer_marketing_df["response"],
    clean_customer_marketing_df["sales_channel"]
)

channel_response_wide_df
```

```python
channel_response_long_df = (
    channel_response_wide_df
    .reset_index()
    .melt(
        id_vars="response",
        var_name="sales_channel",
        value_name="number_of_customers"
    )
)

channel_response_long_df
```

```python
channel_response_long_df["total_customers_in_channel"] = (
    channel_response_long_df
    .groupby("sales_channel")["number_of_customers"]
    .transform("sum")
)
```

```python
channel_response_long_df["response_rate_percent"] = (
    channel_response_long_df["number_of_customers"]
    / channel_response_long_df["total_customers_in_channel"]
    * 100
).round(2)
```

```python
yes_response_rate_by_channel_df = (
    channel_response_long_df[
        channel_response_long_df["response"] == "Yes"
    ]
    [
        [
            "sales_channel",
            "number_of_customers",
            "total_customers_in_channel",
            "response_rate_percent"
        ]
    ]
    .sort_values(
        by="response_rate_percent",
        ascending=False
    )
    .reset_index(drop=True)
)

yes_response_rate_by_channel_df
```
