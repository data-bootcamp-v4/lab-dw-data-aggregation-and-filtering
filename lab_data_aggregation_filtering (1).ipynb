# Lab | Data Aggregation and Filtering

**Dataset:** `marketing_customer_analysis.csv` — Insurance company customer data (demographics, policy details, vehicle information, and response to the last marketing campaign).




## 0. Setup and data loading

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv"
df = pd.read_csv(url)

print(df.shape)
df.head()
```

### Quick column cleanup

The original column names are inconsistent (spaces, mixed case). We standardize them to `snake_case` to make the rest of the analysis easier to write and read, in line with the cleaning we did in the previous lab.

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(' ', '_')
      .str.replace('unnamed:_0', 'id')
)
df.head()
```

```python
df.info()
```

## 1. Customers with low claim amount **and** a "Yes" response

We filter the original DataFrame to keep only customers who:
- have a `total_claim_amount` below \$1,000, **and**
- responded `"Yes"` to the last marketing campaign.


```python
low_claim_responders = df[(df['total_claim_amount'] < 1000) & (df['response'] == 'Yes')]

print(f"Number of matching customers: {len(low_claim_responders)}")
low_claim_responders.head()
```

## 2. Average `monthly_premium_auto` and `customer_lifetime_value` by `policy_type` and `gender` (responders only)

We look only at customers who responded **"Yes"**, group them by `policy_type` and `gender`, and compute the average monthly premium and customer lifetime value. We then compare these figures with the average `total_claim_amount` for the same segments to see which groups look most profitable (high premium/CLV, low claims) or low-risk (low claims).

```python
responders = df[df['response'] == 'Yes']

premium_clv_by_segment = (
    responders
    .groupby(['policy_type', 'gender'])
    .agg(avg_monthly_premium=('monthly_premium_auto', 'mean'),
         avg_customer_lifetime_value=('customer_lifetime_value', 'mean'),
         avg_total_claim_amount=('total_claim_amount', 'mean'),
         n_customers=('customer', 'count'))
    .round(2)
    .sort_values('avg_customer_lifetime_value', ascending=False)
)
premium_clv_by_segment
```

**Interpretation:**

- The segments with the **highest average customer lifetime value** relative to their **average claim amount** are the most profitable: the company collects more value from these customers than it pays out in claims.
- A segment combining a **high monthly premium / high CLV** with a **relatively low average claim amount** is the ideal, low-risk, high-profit customer segment.
- Conversely, segments where the average claim amount is close to (or a large share of) the average premium/CLV represent higher-risk, lower-margin segments for the company.
- In this dataset, `Special Auto` and `Corporate Auto` policyholders tend to show higher average CLV than `Personal Auto` holders, while claim amounts stay comparable across policy types — making the non-personal auto segments relatively more attractive from a profitability standpoint. Differences between genders within the same policy type are generally small.

## 4. Max, min and median customer lifetime value by education level and gender

```python
clv_stats_by_education_gender = (
    df
    .groupby(['education', 'gender'])['customer_lifetime_value']
    .agg(max='max', min='min', median='median')
    .round(2)
)
clv_stats_by_education_gender
```

**Conclusions:**

- The **maximum** customer lifetime value is very high across almost every education/gender combination, indicating that a small number of high-value outlier customers exist in every group (the max values are far above the medians).
- The **median** CLV is much closer across education levels and genders than the maximum, meaning that the "typical" customer's value doesn't vary as dramatically by education as the extremes suggest. Education level alone is not a strong differentiator of typical (median) customer value.
- The **minimum** CLV is low and relatively similar across all groups, showing that every segment includes some low-value customers.
- Because the gap between `max` and `median` is large everywhere, the CLV distribution is **right-skewed** in every education/gender group — this means the *mean* CLV (used in section 2) is pulled upward by a few very high-value customers, and the median is a more robust measure of the "typical" customer in each segment.


## 5. Bonus — Number of policies sold by state and month (pivot table)

We first extract the month from `effective_to_date`, then build a pivot table with **states as rows** and **months as columns**, counting the number of policies sold.

```python
df['effective_to_date'] = pd.to_datetime(df['effective_to_date'])
df['month'] = df['effective_to_date'].dt.month_name()

policies_by_state_month = df.pivot_table(
    index='state',
    columns='month',
    values='policy',
    aggfunc='count',
    fill_value=0
)

# Keep a logical month order (the dataset only covers Jan-Feb 2011)
month_order = [m for m in ['January', 'February', 'March', 'April', 'May', 'June',
                            'July', 'August', 'September', 'October', 'November', 'December']
               if m in policies_by_state_month.columns]
policies_by_state_month = policies_by_state_month[month_order]

policies_by_state_month
```

## 6. Bonus — Top 3 states by number of policies sold, by month

We group by `state` and `month`, count the number of policies sold in each group, identify the **top 3 states** with the highest total number of policies sold, then build a DataFrame showing the monthly breakdown for only those 3 states.

```python
state_month_counts = (
    df.groupby(['state', 'month'])['policy']
    .count()
    .reset_index(name='num_policies_sold')
    .sort_values('num_policies_sold', ascending=False)
)

top_3_states = (
    df.groupby('state')['policy']
    .count()
    .sort_values(ascending=False)
    .head(3)
    .index
    .tolist()
)
print("Top 3 states by number of policies sold:", top_3_states)

top_3_states_monthly = policies_by_state_month.loc[top_3_states]
top_3_states_monthly
```

## 7. Bonus — Customer response rate by marketing channel (melt)

We use `melt` to unpivot the data so that we can compute, for each `sales_channel`, the proportion of customers who responded **"Yes"** to the last marketing campaign.

```python
# Reduce to the two relevant columns and melt
channel_response = df[['sales_channel', 'response']].copy()

melted = channel_response.melt(
    id_vars='sales_channel',
    value_vars='response',
    var_name='campaign',
    value_name='customer_response'
)

response_rate_by_channel = (
    melted
    .assign(is_yes=lambda x: x['customer_response'] == 'Yes')
    .groupby('sales_channel')['is_yes']
    .mean()
    .mul(100)
    .round(2)
    .rename('response_rate_%')
    .sort_values(ascending=False)
)

response_rate_by_channel.to_frame()
```

**Conclusion:** the response rate varies noticeably by marketing channel — channels such as `Agent` tend to convert better than more passive channels like `Web`, which suggests the marketing team should prioritize budget and follow-up effort on the higher-converting channels while investigating why lower-converting channels underperform.
