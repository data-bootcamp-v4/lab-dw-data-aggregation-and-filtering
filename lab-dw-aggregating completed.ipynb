1. Create a new DataFrame that only includes customers who:
   - have a **low total_claim_amount** (e.g., below $1,000),
   - have a response "Yes" to the last marketing campaign.

```python
! pip install pandas
import pandas as pd

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv"
df = pd.read_csv(url)

# Filter: total_claim_amount < 1000 and Response == "Yes"
low_claim_yes_response = df[
    (df['Total Claim Amount'] < 1000) &
    (df['Response'] == 'Yes')
].copy()

print(low_claim_yes_response.shape)
print(low_claim_yes_response.head())
```

2. Using the original Dataframe, analyze:
   - the average `monthly_premium` and/or customer lifetime value by `policy_type` and `gender` for customers who responded "Yes", and
   - compare these insights to `total_claim_amount` patterns, and discuss which segments appear most profitable or low-risk for the company.

```python

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv"
df = pd.read_csv(url)

# Keep only customers who responded "Yes"
df_yes = df[df["Response"] == "Yes"].copy()

# Group by Policy Type and Gender
grouped = (
    df_yes
    .groupby(["Policy Type", "Gender"])[
        ["Monthly Premium Auto", "Customer Lifetime Value", "Total Claim Amount"]
    ]
    .agg("mean")
    .round(2)
)

print(grouped)
```

3. Analyze the total number of customers who have policies in each state, and then filter the results to only include states where there are more than 500 customers.

```python

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv"
df = pd.read_csv(url)

# 1. Count customers per state
customers_by_state = df["State"].value_counts().reset_index()
customers_by_state.columns = ["State", "Customer_Count"]

print("Customers by state (all states):")
print(customers_by_state)

# 2. Filter to states with more than 500 customers
large_states = customers_by_state[customers_by_state["Customer_Count"] > 500].copy()

print("\nStates with more than 500 customers:")
print(large_states)
```

4. Find the maximum, minimum, and median customer lifetime value by education level and gender. Write your conclusions.

```python


url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv"
df = pd.read_csv(url)

# If you have a column named "Customer Lifetime Value" (or similar)
df_grouped = (
    df
    .groupby(["Education", "Gender"])["Customer Lifetime Value"]
    .agg(
        min_value="min",
        max_value="max",
        median_value="median"
    )
    .round(2)
)

print(df_grouped)
```
