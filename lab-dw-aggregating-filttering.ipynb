```python
import pandas as pd

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv"

table_mkt = pd.read_csv(url)

table_mkt.head()
```

```python
condition_1 = table_mkt["Total Claim Amount"] < 1000

condition_2 = table_mkt["Response"] == "Yes"

```

```python
table_mkt_filtered = table_mkt[
    (table_mkt["Total Claim Amount"] < 1000) &
    (table_mkt["Response"] == "Yes")
]
```

```python
#NOS QUEDAMOS SOLO CON LAS COLUMNAS YES

table_mkt_yes = table_mkt[table_mkt["Response"] == "Yes"]
```

```python
#AGRUPAMOS POR DOS COLUMNAS : POLICY TYPE Y GENDER

table_mkt_yes.groupby(["Policy Type", "Gender"])
```

```python
#CALCULAMOS LAS MEDIAS DE LAS TRES COLUMNAS

analysis = table_mkt_yes.groupby(
    ["Policy Type", "Gender"]
)[
    [
        "Monthly Premium Auto",
        "Customer Lifetime Value",
        "Total Claim Amount"
    ]
].mean()
```

```python
analysis = analysis.round(2)

analysis
```

```python
customers_by_state = table_mkt.groupby("State")["Customer"].count()

customers_by_state
```

```python
#FILTRAMOS LOS QUE TIENEN MAS DE 500

customers_by_state[customers_by_state > 500]
```

```python
#Máximo, mínimo y mediana de Customer Lifetime Value
```

```python
table_mkt.groupby(["Education", "Gender"])

table_mkt.groupby(
    ["Education", "Gender"]
)["Customer Lifetime Value"]
```

```python
clv_analysis = table_mkt.groupby(
    ["Education", "Gender"]
)["Customer Lifetime Value"].agg(
    ["max", "min", "median"]
)
```

```python
clv_analysis = clv_analysis.round(2)

clv_analysis
```

```python
#EJERCICIO 5

table_mkt["Effective To Date"] = pd.to_datetime(
    table_mkt["Effective To Date"]
)
```

```python
table_mkt["Month"] = table_mkt["Effective To Date"].dt.month
```

```python
policies_by_state_month = pd.pivot_table(
    table_mkt,
    index="State",
    columns="Month",
    values="Number of Policies",
    aggfunc="sum"
)
```

```python
policies_by_state_month
```

```python
#EJERICIO 6

#PRIMERO AGRUPAMOS POR MES Y ESTADO

policies_state_month = table_mkt.groupby(
    ["State", "Month"]
)["Number of Policies"].sum().reset_index()
```

```python
top_states = (
    policies_state_month
    .groupby("State")["Number of Policies"]
    .sum()
    .sort_values(ascending=False)
    .head(3)
)
```

```python
top_states_names = top_states.index
```

```python
top_3 = policies_state_month[
    policies_state_month["State"].isin(top_states_names)]
```

```python
top_3_table = pd.pivot_table(
    top_3,
    index="State",
    columns="Month",
    values="Number of Policies",
    aggfunc="sum",
    fill_value=0
)

top_3_table
```
